#任务原理解释：

1. **运算符含义**：
   `/` 代表**真除法**，结果永远是浮点数。
   `//` 代表**整除（地板除）**，结果向下取整，返回整数。
   `%` 代表**取模（求余数）**。
   `**` 代表**幂运算**。

2. **类型转换解释**：
   `int("123")` 将纯数字的字符串转换为整数 `123`。
   `float("3.14")` 将字符串转换为浮点数 `3.14`。
   `int(3.9)` 将浮点数强转整数时，直接截断小数部分，不进行四舍五入，因此结果是 `3`。
   `int("goat")` 会抛出 `ValueError`，因为 `"goat"` 是英文字母，不是合法的数学数字，无法转换为整数。

3. **字符串操作**：
   `strip()` 去除了原字符串两端的空格。
   `split()` 默认以空格为分隔符，将字符串切分成一个单词列表。
   `join()` 是 `split()` 的逆操作，用指定的连接符将列表中的单词拼成一个新的长字符串。

4. **字符串不可变性**：
   字符串在 Python 中是不可变对象。这意味着一旦创建，它所在的内存地址的内容就不能被改变。因此 `s1[0] = "h"` 会直接报错。
   要“修改”字符串，实际上是通过切片拼接，如 `"h" + s1[1:]`或者调用 `replace()` 等方法，在内存中生成一个全新的字符串对象。

#实验2：函数、条件、循环与递归

```python
#  任务 1
print("任务1：score_level 函数")

def score_level(score):
    if score < 0 or score > 100:
        raise ValueError(f"成绩 {score} 超出范围，必须在 0~100 之间！")
    
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print(f"95分: {score_level(95)}")
print(f"85分: {score_level(85)}")
print(f"75分: {score_level(75)}")
print(f"55分: {score_level(55)}")

try:
    score_level(105)
except ValueError as e:
    print(f"触发异常{e}")


#  任务 2 
print("\n任务2：for和while的计算")

# for循环
sum_for = 0
for i in range(1, 101):
    sum_for += i
print(f"for 循环结果: {sum_for}")

# while循环
sum_while = 0
j = 1
while j <= 101 - 1:
    sum_while += j
    j += 1
print(f"while 循环结果: {sum_while}")

print(f"两者结果是否相同: {sum_for == sum_while}")


#  任务 3 
print("\n任务3：递归和迭代")

# 递归
def factorial_recursive(n):
    if n < 0:
        raise ValueError("阶乘输入不能为负数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

# 迭代
def factorial_iterative(n):
    if n < 0:
        raise ValueError("阶乘输入不能为负数")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    rec_res = factorial_recursive(n)
    iter_res = factorial_iterative(n)
    print(f"n={n}: 递归结果={rec_res}, 迭代结果={iter_res}, 相同={rec_res == iter_res}")


#  任务 4 
print("\n任务4：summary函数与元组")

def summary(values):
    if not values: 
        raise ValueError("列表不能为空")
    
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    
    return (min_val, max_val, avg_val) 

data = [85, 92, 78, 88, 95]
min_v, max_v, avg_v = summary(data)

print(f"数据: {data}")
print(f"最小值: {min_v}, 最大值: {max_v}, 平均值: {avg_v:.2f}")


#  任务 5 
print("\n任务5：sum_all(*args) 可变参数")

def sum_all(*args):
    return sum(args)

print(f"无参数: {sum_all()}")
print(f"单个参数: {sum_all(10)}")
print(f"多个参数: {sum_all(1, 2, 3, 4, 5)}")
```

#实验5：类、继承与对象复制

```python
import copy

# 任务 1
print("任务1")
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
        
    def is_passed(self):
        return self.score >= 60
        
    def __str__(self):
        return f"学生: {self.name}, 成绩: {self.score}"


# 任务 2
valid_records = [
    {"name": "Alice", "score": 85}, {"name": "Bob", "score": 58},
    {"name": "Carol", "score": 92}, {"name": "David", "score": 76},
    {"name": "Eva", "score": 60}, {"name": "Frank", "score": 89}
]
print("\n任务2")
for rec in valid_records:
    stu = Student(rec["name"], rec["score"])
    status = "及格" if stu.is_passed() else "不及格"
    print(f"{stu}，及格状态: {status}")


# 任务 3 
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score) 
        self.research_topic = research_topic
        
    def __str__(self):
        return f"研究生: {self.name}, 成绩: {self.score}, 研究方向: {self.research_topic}"

print("\n任务3")
grad_stu = GraduateStudent("Grace", 95, "人工智能")
print(grad_stu)


# 任务 4 
print("\n任务4")

original_1 = {"name": "Alice", "scores": [85, 90]}
alias_obj = original_1
alias_obj["scores"].append(95)
print(f"直接赋值修改后 原对象: {original_1}") 


original_2 = {"name": "Alice", "scores": [85, 90]}
shallow_copy = copy.copy(original_2)
shallow_copy["scores"].append(95)
print(f"浅拷贝修改后  原对象: {original_2}")

original_3 = {"name": "Alice", "scores": [85, 90]}
deep_copy = copy.deepcopy(original_3)
deep_copy["scores"].append(95)
print(f"深拷贝修改后  原对象: {original_3}")
```

#任务4 拷贝原理说明：
1. **直接赋值**：本质上是给原对象起了一个新别名（引用），两者在内存中是同一个对象。所以修改任何一方都会影响另一方。
2. **浅拷贝 (`copy.copy`)**：创建了一个新对象，但新对象内部嵌套的可变元素（如列表 `scores`）依然是引用原对象的。所以修改嵌套列表时，原对象会跟着变。
3. **深拷贝 (`copy.deepcopy`)**：不仅创建了新对象，还递归地复制了内部所有的嵌套对象。新对象与原对象在内存中完全独立，无论怎么修改互不影响。
