#欧⼏⾥得算法

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return abs(m) 

print(f"gcd(2018, 2019) = {gcd(2018, 2019)}")
print(f"gcd(1200, 300)  = {gcd(1200, 300)}")
print(f"gcd(5040, 60)   = {gcd(5040, 60)}")
```

#第 3 问解答：
如果函数的参数包含负数，在使用了 `abs()` 函数的情况下，它依然会返回它们最大公约数的绝对值（如 `4`）。
这种行为是有意义的，原因如下：
1. 数学上，最大公约数（GCD）通常被定义为正整数。
2. 从整除的角度来看，如果 `d` 能整除 `m`，那么 `-d` 也能整除 `m`。所以负数参与运算时，算法寻找公共因子的逻辑本身依然成立，最终得出公共因子的最大绝对值，这就是我们想要的 GCD。在代码中使用 `abs()` 保证了输出结果始终符合数学上对 GCD 非负性的严格定义。

#近似欧拉数e

```python
import math

def e_limit(n):
    return (1 + 1 / n) ** n

def e_sum(n):
    if n == 0:
        return 0.0
    
    s = 0.0
    fact = 1.0
    for k in range(n):
        if k > 0:
            fact *= k
        s += 1.0 / fact
    return s

def e_approx(eps):
    real_e = math.exp(1)
    s = 0.0
    k = 0
    fact = 1.0
    
    while abs(s - real_e) > eps:
        if k > 0:
            fact *= k
        s += 1.0 / fact
        k += 1
        
    return s

def print_table(n):
    for k in range(1, n + 1):
        print(e_sum(k))

# 测试
print(f"e_limit(10000)结果: {e_limit(10000)}")

print(f"e_sum(0)结果: {e_sum(0)}")
print(f"e_sum(1)结果: {e_sum(1)}")
print(f"e_sum(4)结果: {e_sum(4)}")

print(f"e_approx(1e-5)结果: {e_approx(1e-5)}")

print_table(5)
```

#测试整数属性

```python
# 任务 1 
def power_1(n):

    while n % 2 == 0 and n > 1:
        n = n // 2
    return n == 1

# 任务 2 
def power_2(b, n):
   
    if b == 1:
        return n == 1
   
    while n % b == 0 and n > 1:
        n = n // b
    return n == 1

#  测试运行 
print("测试 power_1")
print(f"power_1(8):  {power_1(8)}")   
print(f"power_1(1):  {power_1(1)}")   
print(f"power_1(10): {power_1(10)}")  

print("\n测试 power_2")
print(f"power_2(3, 27):   {power_2(3, 27)}")    
print(f"power_2(4, 16):   {power_2(4, 16)}")    
print(f"power_2(2, 10):   {power_2(2, 10)}")    
print(f"power_2(1, 1):    {power_2(1, 1)}")     
print(f"power_2(1, 5):    {power_2(1, 5)}")     
```
